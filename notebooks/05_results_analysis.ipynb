```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Enter the results from Notebook 04

```python
conditions = [
    "Clean",
    "Noise",
    "Blur",
    "Rotation",
    "Occlusion"
]

cnn_results = [
    cnn_clean,
    cnn_noise,
    cnn_blur,
    cnn_rotation,
    cnn_occlusion
]

svm_results = [
    svm_clean,
    svm_noise,
    svm_blur,
    svm_rotation,
    svm_occlusion
]
```

Create a results table

```python
results_df = pd.DataFrame({
    "Condition": conditions,
    "HOG + SVM Accuracy": svm_results,
    "CNN Accuracy": cnn_results
})

results_df
```

Calculate accuracy drop

```python
svm_clean_accuracy = svm_results[0]
cnn_clean_accuracy = cnn_results[0]

svm_drop = [
    svm_clean_accuracy - x
    for x in svm_results
]

cnn_drop = [
    cnn_clean_accuracy - x
    for x in cnn_results
]

results_df["HOG + SVM Accuracy Drop"] = svm_drop
results_df["CNN Accuracy Drop"] = cnn_drop

results_df
```

Compare accuracies visually

```python
x = np.arange(len(conditions))
width = 0.35

plt.figure(figsize=(10, 5))

plt.bar(
    x - width / 2,
    svm_results,
    width,
    label="HOG + SVM"
)

plt.bar(
    x + width / 2,
    cnn_results,
    width,
    label="CNN"
)

plt.xticks(x, conditions)

plt.ylabel("Accuracy")
plt.xlabel("Image Condition")

plt.title(
    "Classical vs Deep Vision: Accuracy Under Image Degradation"
)

plt.legend()

plt.tight_layout()
plt.show()
```

Plot robustness drop

```python
plt.figure(figsize=(10, 5))

plt.plot(
    conditions,
    svm_drop,
    marker="o",
    label="HOG + SVM"
)

plt.plot(
    conditions,
    cnn_drop,
    marker="o",
    label="CNN"
)

plt.ylabel("Accuracy Drop")
plt.xlabel("Image Condition")

plt.title(
    "Performance Degradation Relative to Clean Images"
)

plt.legend()

plt.tight_layout()
plt.show()
```

Determine the better model under each condition

```python
for condition, svm_acc, cnn_acc in zip(
    conditions,
    svm_results,
    cnn_results
):

    if cnn_acc > svm_acc:
        winner = "CNN"

    elif svm_acc > cnn_acc:
        winner = "HOG + SVM"

    else:
        winner = "Equal"

    print(
        f"{condition}: {winner}"
    )
```

Calculate average robustness

```python
average_svm_accuracy = np.mean(svm_results)
average_cnn_accuracy = np.mean(cnn_results)

print(
    "Average HOG + SVM Accuracy:",
    round(average_svm_accuracy, 4)
)

print(
    "Average CNN Accuracy:",
    round(average_cnn_accuracy, 4)
)
```

Find the most damaging degradation

```python
degradation_conditions = conditions[1:]

svm_degradation_results = svm_results[1:]
cnn_degradation_results = cnn_results[1:]

worst_svm_index = np.argmin(
    svm_degradation_results
)

worst_cnn_index = np.argmin(
    cnn_degradation_results
)

print(
    "Most difficult degradation for HOG + SVM:",
    degradation_conditions[worst_svm_index]
)

print(
    "Most difficult degradation for CNN:",
    degradation_conditions[worst_cnn_index]
)
```

Final interpretation section

## Experimental Interpretation

This experiment compares a classical HOG + SVM pipeline with a convolutional neural network under clean and degraded visual conditions.

The analysis focuses on two questions:

1. Which method achieves stronger classification performance on clean images?
2. Which method maintains its performance more effectively when images are corrupted by noise, blur, rotation, and occlusion?

The results are interpreted using both absolute accuracy and the reduction in performance relative to clean images.

The experiment is intended as an educational study of the transition from handcrafted visual representations to learned deep representations and should not be interpreted as a claim of a novel robustness method.
